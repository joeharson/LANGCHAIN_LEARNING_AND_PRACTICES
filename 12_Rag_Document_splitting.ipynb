### 1) Token based

In this charecter text Splitter , when we have 220  token and we assign chunk size to 200, but still it split 220 and not force split to 200 , this is CharecterTextSplitter . but TokenTextSplitter is force to split 200 token given in chunk size  

```python
import tiktoken

# This works without any OpenAI API key
encoding = tiktoken.get_encoding("cl100k_base")
tokens = encoding.encode("Hello world!")

print(f"Tokens: {tokens}")
print(f"Token count: {len(tokens)}")
```

```python
from langchain_text_splitters import CharacterTextSplitter 

text_splitter = CharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base",
    chunk_size=200,
    chunk_overlap=0
)

# cl100k_base -> GPT-3.5-turbo, GPT-4, GPT-4-turbo
# p50k_base -> Codex models, text-davinci-002/003, -> gpt 3 family
# r50k_base -> GPT-3 models (davinci, curie, etc.)

texts = text_splitter.split_documents(documents)

print(f"Length: {len(texts)}")
texts
```

```python
from langchain_text_splitters import TokenTextSplitter 

text_splitter = TokenTextSplitter(
    encoding_name="cl100k_base",
    chunk_size=200,
    chunk_overlap=40
)
texts = text_splitter.split_documents(documents)
print(f"Length: {len(texts)}")
texts
```

### Document Splitter

Document based splitting
- MD .split_text()
- JSON .split_json()
- HTML .split_text()
- Code .create_documents()

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

markdown_document = """# LangChain
## What is it?
A framework to **build apps with LLMs** – think AI meets Lego blocks.

### Core idea
Combine **prompts**, **chains**, and **agents** to make smart workflows.

#### Example
`Translate: "Hello" -> "Bonjour"`
"""

headers_to_split_on = [
    ("#", "Primary header"),
    ("##", "Secondary header"),
    ("###", "Tertiary header"),
]

markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on,
    strip_headers=False,   # when give True it will remove the headers from the text, when False it will keep the headers in the text
    return_each_line=True
)
md_header_splits = markdown_splitter.split_text(markdown_document)

print(f"Length: {len(md_header_splits)}")
md_header_splits
```

Add RecursiveCharacterTextSplitter on top of MarkdownHeaderTextSplitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter 

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)
texts = text_splitter.split_documents(md_header_splits)
print(f"Length: {len(texts)}")
texts
```

```python
import json
from langchain_text_splitters import RecursiveJsonSplitter
from pprint import pprint

# --- JSON Data Definition 1 (Object with a nested 'cars' list) ---
json_data_1 = """{
    "cars":[
        {
            "make": "Maruti Suzuki",
            "model": "Baleno",
            "year": 2023,
            "price_in_inr": 900000,
            "fuel_type": "Petrol"
        },
        {
            "make": "Tata",
            "model": "Punch",
            "year": 2024,
            "price_in_inr": 750000,
            "fuel_type": "Petrol"
        },
        {
            "make": "Mahindra",
            "model": "XUV700",
            "year": 2023,
            "price_in_inr": 2000000,
            "fuel_type": "Diesel"
        }
    ]
}"""

# --- JSON Data Definition 2 (Direct list of objects) ---
json_data_2 = """[
    {
        "make": "Maruti Suzuki",
        "model": "Baleno",
        "year": 2023,
        "price_in_inr": 900000,
        "fuel_type": "Petrol"
    },
    {
        "make": "Tata",
        "model": "Punch",
        "year": 2024,
        "price_in_inr": 750000,
        "fuel_type": "Petrol"
    },
    {
        "make": "Mahindra",
        "model": "XUV700",
        "year": 2023,
        "price_in_inr": 2000000,
        "fuel_type": "Diesel"
    }
]"""

# --- Splitting Logic ---
# Load the raw string into a Python dictionary/list
json_data = json.loads(json_data_1)

# Initialize the text splitter with chunk size parameters
splitter = RecursiveJsonSplitter(max_chunk_size=200)

# Recursively split json data - If you need to access/manipulate the smaller json chunks
json_chunks = splitter.split_json(json_data=json_data, convert_lists=True)  # here if give False then it will not split the list of objects, it will keep the list as a single chunk


print(f"Length: {len(json_chunks)}")

# Display the output chunks
json_chunks
```

this is similar to markdowntextsplitter

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

# --- HTML Data Definition ---
html_data = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>HTML Text Splitter</title>

    <style>
        body {
            font-family: Arial, sans-serif;
            max-width: 1000px;
            margin: 40px auto;
            padding: 20px;
        }

        textarea {
            width: 100%;
            height: 250px;
            padding: 12px;
            font-size: 15px;
            margin-bottom: 15px;
        }

        button {
            padding: 10px 20px;
            font-size: 16px;
            cursor: pointer;
        }

        .chunk {
            border: 1px solid #ccc;
            padding: 15px;
            margin-top: 15px;
            border-radius: 8px;
        }

        .chunk-title {
            font-weight: bold;
            margin-bottom: 10px;
        }

        .content {
            background: #f5f5f5;
            padding: 10px;
            white-space: pre-wrap;
        }
    </style>
</head>

<body>

    <h1>HTML Text Splitter</h1>

    <p>
        Paste your HTML below and split it based on HTML headings.
    </p>

    <textarea id="htmlInput">
<h1>LangChain</h1>

<p>LangChain is a framework for building applications with LLMs.</p>

<h2>Text Splitters</h2>

<p>Text splitters divide large documents into smaller chunks.</p>

<h3>Recursive Character Splitter</h3>

<p>RecursiveCharacterTextSplitter tries to preserve meaningful text.</p>

<h2>RAG</h2>

<p>RAG allows an LLM to retrieve information from documents.</p>
    </textarea>

    <br>

    <button onclick="splitHTML()">Split HTML</button>

    <div id="output"></div>


    <script>

        function splitHTML() {

            const html = document.getElementById("htmlInput").value;

            const parser = new DOMParser();

            const documentHTML =
                parser.parseFromString(html, "text/html");

            const elements =
                documentHTML.body.querySelectorAll("h1, h2, h3, h4, h5, h6, p");

            const chunks = [];

            let currentChunk = {
                heading: "",
                content: []
            };


            elements.forEach(element => {

                const tag = element.tagName.toLowerCase();

                // If heading → start new chunk
                if (tag.startsWith("h")) {

                    if (currentChunk.content.length > 0) {

                        chunks.push({
                            heading: currentChunk.heading,
                            content: currentChunk.content.join("\n")
                        });

                    }

                    currentChunk = {
                        heading: element.innerText,
                        content: []
                    };

                }

                // If paragraph → add to current chunk
                else {

                    currentChunk.content.push(
                        element.innerText
                    );

                }

            });


            // Add final chunk
            if (currentChunk.content.length > 0) {

                chunks.push({
                    heading: currentChunk.heading,
                    content: currentChunk.content.join("\n")
                });

            }


            // Display results

            const output =
                document.getElementById("output");

            output.innerHTML = "";


            chunks.forEach((chunk, index) => {

                const div =
                    document.createElement("div");

                div.className = "chunk";

                div.innerHTML = `
                    <div class="chunk-title">
                        Chunk ${index + 1}
                    </div>

                    <strong>
                        ${chunk.heading}
                    </strong>

                    <div class="content">
                        ${chunk.content}
                    </div>
                `;

                output.appendChild(div);

            });

        }

    </script>

</body>
</html>
"""

# --- Splitter Setup and Execution ---
headers_to_split_on = [
    ("h1", "Primary header"),
    ("h2", "Secondary header"),
    ("h3", "Tertiary header")
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)

html_chunks = html_splitter.split_text(html_data)

print(f"Length of html splitter: {len(html_chunks)}")

# --- Printing Results ---
from pprint import pprint

for chunk in html_chunks:
    pprint(f"Metadata: {chunk.metadata}")
    pprint(f"Chunk.{chunk.page_content}")
    print("\n")

### Further splitting HTML header chunks with RecursiveCharacterTextSplitter

print("FURTHER SPLITTING BASED ON CHARACTER", "\n\n\n")
```

```python
from langchain_text_splitters import Language, RecursiveCharacterTextSplitter

# # for language in Language:
# #     print(language.value)

# # print(*[lang.value for lang in Language], sep="\n")

sample_python_code = """
def is_palindrome(s):
    s = s.lower().replace(" ", "")  # ignore case and spaces
    return s == s[::-1]             # reverse string and compare

# Test
words = ["level", "Racecar", "hello", "madam"]
for word in words:
    print(word, "->", is_palindrome(word))
"""

python_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=200,
    chunk_overlap=30
)

python_chunks = python_splitter.create_documents([sample_python_code],[{"file": "palindrome.py"}])  # [{"file": "palindrome.py"}] this is metadata for the document, you can add any metadata you want

print(f"Length of python chunks: {len(python_chunks)}")
for chunk in python_chunks:
    print(f"Metadata: {chunk.metadata}")
    print(f"{chunk.page_content}")
    print("\n")


## spliting like Text
# # python_chunks = python_splitter.split_text(sample_python_code)
# # print(f"Length of python chunks: {len(python_chunks)}")
# # for chunk in python_chunks:
# #     print(chunk)
# #     print("\n")
```
