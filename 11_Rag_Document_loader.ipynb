### Document Loader

#### Langchain Loader reference : https://docs.langchain.com/oss/python/integrations/document_loaders

```python
import langchain 
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate , ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()
Groq_api_key=os.getenv("GROQ_API_KEY")

llm=ChatGroq(
    api_key=Groq_api_key,
    model="openai/gpt-oss-120b",
    temperature=0.2
    )

```

Text Loader

```python
from langchain_community.document_loaders import TextLoader

file_path = r"./Additional files/10_vijay_Small_RAG_Content.txt"

loader = TextLoader(file_path, encoding="utf-8")

documents = loader.load()

print(documents[0].page_content)
```

CSV Loader

```python
from langchain_community.document_loaders.csv_loader import CSVLoader

file_path = r"./Additional files/customers_50.csv"

loader = CSVLoader(file_path)

data = loader.load()
for document in data:
    print(documents[0].page_content,"\n")
```

WebBaseLoader

With lazy Loader , it return immediately without waiting for loading all datas 

```python
from langchain_community.document_loaders import WebBaseLoader

file_path = r"https://www.britannica.com/biography/Jesus"

#load single website

# loader = WebBaseLoader(file_path)
# data = loader.load()
# for document in data:
#     print(document.page_content,"\n")



# Load Multiple Website
multiple_websites_loader = WebBaseLoader(
    [
    "https://www.britannica.com/biography/Jesus",
    "https://www.britannica.com/biography/Mary-mother-of-Jesus",
    "https://www.britannica.com/biography/St-James-son-of-Zebedee"
    ]
)

docs=multiple_websites_loader.lazy_load()
for document in docs:
    print(document.page_content,"\n")
```
